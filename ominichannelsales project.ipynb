```python
store_summary = df.groupby("Store ID").agg({
    "revenue": "sum",
    "Total Cost": "sum",
    "profit": "sum",
    "Units Sold": "sum"
}).reset_index()

print(store_summary.head())
```

```python
import matplotlib.pyplot as plt

store_summary = store_summary.sort_values(
    by="revenue",
    ascending=False
)

plt.figure(figsize=(12,6))

plt.bar(
    store_summary["Store ID"],
    store_summary["revenue"]
)

plt.title("Revenue by Store")
plt.xlabel("Store")
plt.ylabel("Revenue")

plt.xticks(rotation=90)

plt.show()
```

```python
plt.figure(figsize=(12,6))

plt.bar(
    store_summary["Store ID"],
    store_summary["profit"]
)

plt.title("Profit by Store")
plt.xlabel("Store")
plt.ylabel("Profit")

plt.xticks(rotation=90)

plt.show()
```
